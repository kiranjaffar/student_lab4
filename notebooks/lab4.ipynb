```python
import numpy as np

A = np.array([[1, 2],
              [3, 4]])

B = np.array([[10, 20],
              [30, 40]])

print("Element-wise multiplication:")
print(A * B)

print("Matrix multiplication:")
print(A @ B)

print("Transpose of A:")
print(A.T)
```

```python
M = np.array([[2, 1],
              [1, 3]])

rhs = np.array([5, 10])

sol = np.linalg.solve(M, rhs)

print("Solution [x, y]:", sol)
print("Determinant:", np.linalg.det(M))
print("Solution verified:", np.allclose(M @ sol, rhs))
print("Inverse of M:")
print(np.linalg.inv(M))

print("Norm of [3, 4]:", np.linalg.norm(np.array([3, 4])))
```

```python
# Part 6: Indexing and Slicing

import numpy as np

scores = np.array([
    [45, 78, 90, 60],
    [32, 65, 70, 55],
    [88, 92, 79, 95]
])

print("First student, second subject:")
print(scores[0, 1])

print("Last student, second-last subject:")
print(scores[-1, -2])

print("First subject for all students:")
print(scores[:, 0])

print("First two students, middle two subjects:")
print(scores[0:2, 1:3])
```

```python
print(scores[-1, 0:2])
print(scores[0:2, 2])
print(scores[0:2, 2:4])
```

```python
import numpy as np

scores = np.array([
    [45, 78, 90, 60],
    [32, 65, 70, 55],
    [88, 92, 79, 95]
])

# Add a different bonus to each subject
bonus = np.array([5, 0, 10, 0])

print("Scores after subject bonuses:")
print(scores + bonus)

# Broadcasting a column and a row
col = np.array([[1], [2], [3]])
row = np.array([10, 20, 30])

print("Column plus row:")
print(col + row)

# Subtract each subject's mean from its scores
means = scores.mean(axis=0)

print("Subject means:")
print(means)

print("Scores centred around subject means:")
print(np.round(scores - means, 1))
```

```python
import numpy as np

marks_list = [40, 60, 72]

# Using a Python loop
with_loop = []
for mark in marks_list:
    with_loop.append(mark / 80 * 100)

# Using NumPy vectorization
vectorized = np.array(marks_list) / 80 * 100

print("Loop result:", with_loop)
print("NumPy result:", vectorized)
```

```python
scores = np.array([
    [45, 78, 90, 60],
    [32, 65, 70, 55],
    [88, 92, 79, 95]
])

print("Total:", scores.sum())
print("Mean:", scores.mean())
print("Minimum:", scores.min())
print("Maximum:", scores.max())

print("Sum of each subject:", scores.sum(axis=0))
print("Sum of each student:", scores.sum(axis=1))

print("Population standard deviation:", round(scores.std(), 2))
print("Sample standard deviation:", round(scores.std(ddof=1), 2))

print("Scores greater than or equal to 60:")
print(scores >= 60)
```

```python
# Student averages
print(scores.mean(axis=1))

# Highest score in each subject
print(scores.max(axis=0))

# Number of scores >= 60
print(np.sum(scores >= 60))
```
